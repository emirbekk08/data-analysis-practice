```python
plt.figure(figsize=(8, 5))

plt.hist(df["total_bill"], bins=20, edgecolor="black")

plt.title("Распределение суммы счета")
plt.xlabel("Сумма счета")
plt.ylabel("Количество посетителей")

plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
tips_by_day = df.groupby("day")["tip"].mean()

plt.figure(figsize=(8, 5))

tips_by_day.plot(kind="bar", edgecolor="black")

plt.title("Средний размер чаевых по дням недели")
plt.xlabel("День недели")
plt.ylabel("Средние чаевые")

plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

## Наблюдения по результатам анализа

1. **Распределение счетов:** суммы счетов распределены неравномерно. Большинство наблюдений сосредоточено в диапазоне сравнительно небольших и средних счетов, а крупные счета встречаются реже.

2. **Чаевые по дням недели:** средний размер чаевых отличается в зависимости от дня недели. Для точного сравнения необходимо учитывать значения, показанные на втором графике.

3. **Связь суммы счета и чаевых:** размер чаевых может зависеть от суммы счета. Эту зависимость можно дополнительно проверить с помощью диаграммы рассеяния.

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["total_bill"], df["tip"], alpha=0.7)

plt.title("Зависимость чаевых от суммы счета")
plt.xlabel("Сумма счета")
plt.ylabel("Размер чаевых")

plt.grid(alpha=0.3)
plt.show()
```
